Automatic File Sorter In File Explorer

```python
import os, shutil
```

```python
path = r"C:/Users/DELL/OneDrive/Pictures/PYTHONPROJUSE7/"
```

Seeing what files are there in the  folder

```python
 file_name = os.listdir(path)
```

To check if the folder/path  exists or not

```python
os.path.exists(path)
```

To check if these separate paths/folders are created or not


```python
folder_names = ['csv files','image files','text files']
for loop in range(0,2):
    os.path.exists(path + folder_names[loop])
               
```

If not created 
Now actually creating


```python
folder_names = ['csv files','image files','text files']
for loop in range(0,3):
    if not os.path.exists(path + folder_names[loop]):
         os.makedirs((path + folder_names[loop]))
        
```

Writing code to check file type and automatically sort in its types file

```python
for file in file_name:
    if ".csv" in file and not os.path.exists(path + "csv files/" +file):
         shutil.move(path + file, path + "csv files/" + file)
    elif ".png" in file and not os.path.exists(path + "image files/" +file):
        shutil.move(path + file, path + "image files/" + file)
    elif ".txt" in file and not os.path.exists(path + "text files/" +file):
        shutil.move(path + file, path + "text files/" + file)
   
```
