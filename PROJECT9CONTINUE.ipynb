# FLIPKART WEB SCRAPING THROUGH ALL/MULTIPLE PAGES

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

import warnings
warnings.filterwarnings("ignore")

import requests
from bs4 import BeautifulSoup
from urllib.request import urlopen, Request

url_list = []
for i in range(1,42):
     url = "https://www.flipkart.com/search?q=mobile&otracker=search&otracker1=search&marketplace=FLIPKART&as-show=on&as=off"
     url_list.append(url)
    
def create_html_page(url):
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                      "AppleWebKit/537.36 (KHTML, like Gecko) "
                      "Chrome/136.0.0.0 Safari/537.36"
    }
    response = requests.get(url, headers=headers)

    # Check if request succeeded
    if response.status_code == 200:
        page_soup = BeautifulSoup(response.text, "html.parser")
        return page_soup
    else:
        print(f"Error: {response.status_code}")
        return None
```

```python


def create_final_data(page_soup):
    data = {
        "ProductName": [],
        "Review&Ratings": [],
        "RatingNumber": [],
        "Memory": [],
        "Camera": [],
        "Processor": [],
        "Warranty": [],
        "Battery": [],
        "Sale & Actual Price": [],
        "Offer": []
    }
    all_info_class_name = "ZFwe0M row"
    containers = page_soup.find_all("div",{"class":all_info_class_name})
    for contain in containers:
        # Product name
        product_name = contain.find("div", {"class": "RG5Slk"})
        data["ProductName"].append(product_name.text if product_name else None)

        # Reviews & ratings
        rr = contain.find("span", {"class": "PvbNMB"})
        if rr:
            data["Review&Ratings"].append(rr.text)
            rating_tag = contain.find("div", {"class": "MKiFS6"})
            data["RatingNumber"].append(rating_tag.text if rating_tag else None)
        else:
            data["Review&Ratings"].append(None)
            data["RatingNumber"].append(None)

        # Specifications
        spec_list = contain.find("ul", {"class": "HwRTzP"})
        if spec_list:
            lis = spec_list.find_all("li")
            data["Memory"].append(lis[0].text if len(lis) > 0 else None)
            data["Camera"].append(lis[2].text if len(lis) > 2 else None)
            data["Processor"].append(lis[4].text if len(lis) > 4 else None)
            data["Warranty"].append(lis[5].text if len(lis) > 5 else None)
            data["Battery"].append(lis[3].text if len(lis) > 3 else None)
        else:
            data["Memory"].append(None)
            data["Camera"].append(None)
            data["Processor"].append(None)
            data["Warranty"].append(None)
            data["Battery"].append(None)

        # Price
        price = contain.find("div", {"class": "QiMO5r"})
        data["Sale & Actual Price"].append(price.text if price else None)

        # Offer
        offer_list = contain.find("div", {"class": "HQe8jr"})
        data["Offer"].append(offer_list.text if offer_list else "No Offer")

    # Build DataFrame after loop
    df = pd.DataFrame(data)
    print(df.head())

    return data
```

# MAIN SECTION

```python
data_frame_list = []
for i,url in enumerate(url_list):
    print(f"Started url page {i+1}")
    page_soup = create_html_page(url_list[i])
    data_info = create_final_data(page_soup)
    data_frame_list.append(data_info)
```
