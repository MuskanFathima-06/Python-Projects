```python
                                                  #BMI CALCULATOR    

name = input("Enter Your Name: ")
if not name.isalpha():
    raise ValueError("Name must contain only letters")
else:
     weight = int(input("Enter Your Weight In Pounds:"))
    
     height = int(input("Enter Your Height In Inches:"))
     BMI = (weight * 703) / (height * height)
     print(BMI)
if BMI > 0:
   if(BMI < 18.5):
     print(name +", You Are Underweight and You Have Minimal Health Risk.")
   elif(BMI <= 24.9):
     print(name +", You Have Normalweight and You Have Minimal Health Risk.")
   elif(BMI < 29.9):
     print(name +", You Are Overweight and You Have Increased Health Risk.")
   elif(BMI < 34.9):
     print(name +", You Are Obese and You Have High Health Risk.")
   elif(BMI < 39.9):
     print(name +", You Are Severely Obese and You Have Very High Health Risk.")
   elif(BMI >= 40):
     print(name +", You Are Morbidly Obese and You Have Extremely High Health Risk.")
   else:
      print("Enter Valid Input") 
```
